### Handling Missing Values Numerical

```python
# Fill missing values in numerical columns with the median
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

### Observation - Filtering out the Numerical columns and Categorical Columns

### Fill Missing Values in Categorical Values

```python
# Fill missing values in categorical columns with the mode
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## Outlier Detection and Handling


### We will use the Z-score method to identify outliers in key numerical variables and visualize them using Box Plots.

```python
# Create boxplots to visualize outliers in key numerical columns
import numpy as np
selected_cols = ['age', 'salary_usd', 'experience_years']


plt.figure(figsize=(15, 5))
for i, col in enumerate(selected_cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

```python
# Detect outliers using Z-score method (threshold > 3)
z_scores = (df[selected_cols] - df[selected_cols].mean()) / df[selected_cols].std()
outliers_mask = (np.abs(z_scores) > 3).any(axis=1)

print(f"Total rows with outliers (Z-score > 3): {outliers_mask.sum()}")

if outliers_mask.any():
    display(df[outliers_mask].head())
```

```python
# Detect outliers again and remove them into a clean variable
z_scores = (df[selected_cols] - df[selected_cols].mean()) / df[selected_cols].std()
outliers_mask = (np.abs(z_scores) > 3).any(axis=1)
print("Outliers detected:", outliers_mask.sum())

# df_clean = dataframe with all outliers completely removed
df_clean = df[~outliers_mask].reset_index(drop=True)
print("Rows after removing outliers:", len(df_clean))
```

#### *   Detection Methodology: Used Box Plots for visual identification and the Z-Score method (threshold > 3) to statistically flag 256 rows as significant outliers in age, salary, and experience.

#### *   Validation: Confirmed that these outliers represent legitimate extreme values in the dataset rather than data entry errors, ensuring the integrity of the subsequent analysis.

### Q1: Which industries achieve the highest average productivity improvement after adopting AI?

```python
# Map productivity change to numeric values and compute average by industry
mapping = {'Increased': 1, 'Same': 0, 'Decreased': -1}
df_clean['productivity_numeric'] = df_clean['productivity_change'].map(mapping)

industry_productivity = df_clean.groupby('industry')['productivity_numeric'].mean().sort_values(ascending=False)

industry_productivity
```

```python
# Plot average productivity improvement by industry
plt.figure(figsize=(12,6))

sns.barplot(
    x=industry_productivity.values,
    y=industry_productivity.index,
    hue=industry_productivity.index,
    palette='Set2',
    legend=False
)

plt.title("Average Productivity Improvement by Industry")
plt.xlabel("Average Productivity Change")
plt.ylabel("Industry")

plt.show()
```

**Observation:** This code computes the average productivity improvement score for each industry and shows it in a bar chart.

### Q4. Does the average salary differ across company sizes?

```python
# Compute average salary grouped by company size
df_clean.groupby("company_size")["salary_usd"].mean()
```

```python
# Calculate IQR for salary_usd to identify and remove 'normal' outliers
Q1 = df_clean['salary_usd'].quantile(0.25)
Q3 = df_clean['salary_usd'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Create a new DataFrame without these IQR-based outliers for salary_usd
df_q4_cleaned = df_clean[(df_clean['salary_usd'] >= lower_bound) & (df_clean['salary_usd'] <= upper_bound)]

print(f"Original rows in df_clean: {len(df_clean)}")
print(f"Rows after removing IQR outliers for salary_usd: {len(df_q4_cleaned)}")
```

A new DataFrame `df_q4_cleaned` has been created, filtering out additional outliers from `salary_usd` based on the 1.5 * IQR rule. The boxplot below will now use this refined dataset.

```python
# Boxplot comparing salary distribution across company sizes
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df_q4_cleaned, # Changed from df_clean to df_q4_cleaned
    x="company_size",
    y="salary_usd"
)

plt.title("Salary Distribution Across Company Sizes (IQR Outliers Removed)")
plt.xlabel("Company Size")
plt.ylabel("Salary (USD)")

plt.show()
```

**Observation:** This code computes the average salary for each company size and shows the distribution in a boxplot.
